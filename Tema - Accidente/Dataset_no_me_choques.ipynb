# Dataset: Accidente
---

Cada fila representa un evento de accidente víal, cada columna contiene atributos de este. 

La columna *sevrity* indica la severidad del accidente

# Variables

Variable | Descripción
:----    | :----
ID | Identificador único
Source | Fuente/origen del reporte
TMC | Código "Traffic Message Channel" o TMC
Severity | Severidad del accidente, [1, 2, 3, 4], donde 1 es el más leve
Start_Time | Hora de inicio
End_Time | Hora fin
Start_Lat | Latitud GPS inicial
Start_Lng | Longitud GPS inicial
End_Lat | Latitud GPS final
End_Lng | Longitud GPS final
Distance(mi) | Longitud del camino afectado por el accidente
Description | Descripción natural
Number | Número de calle
Street | Calle
Side | Dirección de la calle (Right/Left)
City | Ciudad
County | Condado
State | Estado
Zipcode | Código postal
Country | País
Timezone | Zona horaria (eastern, central, etc.).
Airport_Code | Código aeroportuario más cercano al accidente
Weather_Timestamp | time-stamp del clima.
Temperature(F) | Temperatura (Fahrenheit).
Wind_Chill(F) | Viento (Fahrenheit).
Humidity(%) | Humedad relativa (porcentaje).
Pressure(in) | Presión atmosférica (pulgadas).
Visibility(mi) | Visibilidad (millas).
Wind_Direction | Dirección del viento.
Wind_Speed(mph) | Velocidad del viento (milla x Hora).
Precipitation(in) | Cantidad de precipitaciones, si las hay
Weather_Condition | Condición atmosférica (rain, snow, thunderstorm, fog, etc.).
Amenity | Punto de interes (PdI) que indica la presencia de una amenity cerca
Bump | Punto de interes (PdI) que indica la presencia de un reductor de velocidad cerca
Crossing | Punto de interés (PdI) que indica la presencia de una senda peatonal cerca
Give_Way | Punto de interés (PdI) que indica la presencia de un signo de alejarse
Junction | Punto de interés (PdI) que indica la presencia de una intersección cerca
No_Exit | Punto de interés (PdI) que indica la presencia de una calle sin salida cerca
Railway | Punto de interés (PdI) que indica la presencia de un cruce de tren cerca
Roundabout | Punto de interés (PdI) que indica la presencia de una ruta provincial cerca
Station | Punto de interés (PdI) que indica la presencia de una estación (tren, autobus, etc) cerca
Stop | Punto de interés (PdI) que indica la presencia de un signo de alto cerca
Traffic_Calming | Punto de interés (PdI) que indica la presencia de una señal de tráfico calmado cerca
Traffic_Signal | Punto de interés (PdI) que indica la presencia de una señal de mucho tráfico cerca
Turning_Loop | Punto de interés (PdI) que indica la presencia de una señal de giro en u cerca
Sunrise_Sunset | Período del dia (day, night) según la salida o caída del sol
Civil_Twilight | Período del dia (day, night) según el crepúesculo local
Nautical_Twilight | Período del dia (day, night) según el crepúesculo naútico.
Astronomical_Twilight | Período del dia (day, night) según el crepúesculo astronómico.
